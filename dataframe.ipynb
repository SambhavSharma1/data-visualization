```python
import pandas as pd
import numpy as np
```

```python
student_df = pd.DataFrame({
    'Name': ['Aman', 'Riya', 'Rahul', 'Priya'],
    'Age': [20, 19, 21, 20],
    'City': ['Delhi', 'Mumbai', 'Mathura', 'Agra']})
print(student_df)
```

```python
print(student_df)
```

```python
print(student_df.columns)
```

```python
print(student_df['Name'])
```

```python
print(student_df['City'])
```

```python
new_df = student_df[['Name', 'City']]
print(new_df)
```

```python
book_df = pd.DataFrame({
    'Book': ['Python', 'Java', 'DBMS'],
    'Author': ['Amit', 'Rahul', 'Neha'],
    'Price': [400, 500, 350]})
print(book_df)
```

```python
book_df['Publisher'] = 'ABC Publications'
print(book_df)
```

```python
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D', 'E']})
df['SerialNo'] = np.arange(5)
print(df)
```

```python
df = df.drop('SerialNo', axis=1)
print(df)
```

```python
movie_df = pd.DataFrame({
    'Movie': ['3 Idiots', 'Dangal', 'PK'],
    'Year': [2009, 2016, 2014],
    'Rating': [8.4, 8.3, 8.1]})
print(movie_df)
```

```python
new_movie_df = movie_df[['Movie', 'Rating']].copy()
new_movie_df['Director'] = np.nan
print(new_movie_df)
```

```python
cities = pd.Series(
    ['Delhi', 'Mumbai'],
    index=[0, 3])

print(cities)
```

```python
employee_df = pd.DataFrame({
    'Employee': ['Aman', 'sambhav', 'sumit', 'sachin', 'Karan'],
    'Department': ['IT', 'HR', 'Sales', 'IT', 'Finance'],
    'Experience': [2, 4, 3, 5, 1]})
new_employee_df = employee_df[['Employee', 'Experience']]
print(new_employee_df)
```

```python
print(employee_df['Department'])
print(employee_df.columns)
```

```python
employee_df['Office'] = 'Head Office'
print(employee_df)
```

```python
employee_df['Office'] = np.arange(5)
print(employee_df)
```

```python
office_names = pd.Series(
    ['Delhi', 'Mumbai', 'Pune'],
    index=[0, 2, 4])
employee_df['Office'] = office_names
print(employee_df)
```

```python
employee_df = employee_df.drop('Office', axis=1)
print(employee_df)
```

```python
data = {
    'Name': ['Aman', 'sambhav', 'rudrakshi'],
    'Age': [20, 19, 21],
    'City': ['Delhi', 'Agra', 'Mathura']}
df = pd.DataFrame(data,columns=['Name', 'Age', 'City', 'Marks'])
print(df)
```

```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Rahul', 'Neha', 'Karan'],
    'Course': ['Python', 'Java', 'DBMS', 'Python', 'Java'],
    'Score': [85, 90, 78, 88, 82]})
print(course_df.columns)
print(course_df['Student'])
new_df = course_df[['Student', 'Score']]
print(new_df)
course_df['Batch'] = '2026'
print(course_df)
```

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Book', 'Bag', 'Pencil'],
    'Quantity': [50, 30, 10, 40],
    'Price': [10, 100, 500, 5]})
stock = pd.Series([100, 200],index=[0, 2])
inventory_df['Stock'] = stock
print(inventory_df)
inventory_df = inventory_df.drop('Stock',axis=1)
print(inventory_df)
```
